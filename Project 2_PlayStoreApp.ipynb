**Handling NaN / Null / Missing Values**

Data Cleaning Process:

Removing duplicate records

Handling missing values (NaN)

```python
#Checks how many null values are present in each column
apps.isnull().sum()
```

```python
reviews.isnull().sum()
```

**Remove / Handle Missing Values (Apps Dataset)**

Handling Missing Values:

Dropping rows where Rating is missing

```python
# Remove rows where Rating is missing
apps = apps[apps["Rating"].notna()]
```

```python
# Optional: remove all null rows
apps = apps.dropna()
```

**Handle Missing Values (Reviews Dataset)**

```python
# Remove rows where sentiment is missing
reviews = reviews.dropna(subset=["Sentiment"])
```

```python
# Optional: remove all null rows
reviews = reviews.dropna()
```

```python
#Verify again
reviews.isnull().sum()
apps.isnull().sum()
```

#**9) Conclusions from Data,Trends**


**Aggregation means combining data to get a summary value**

Instead of looking at individual rows, you group data and calculate totals, averages, etc.

***`🔹 Aggregation & Trends`***

```python
#I used aggregation to calculate total installs for each category. This helps identify which app category is most popular based on downloads
apps.groupby("Category")["Installs"].sum().sort_values(ascending=False)

#I calculated average ratings for free and paid apps using aggregation. This helps compare performance between different app types.
apps.groupby("Type")["Rating"].mean()
```

# **10) Visualization**

Visualization means representing data in graphical form (charts, graphs, plots)

**I created multiple visualizations such as:**

***Histogram***→ to analyze rating distribution

***Bar charts*** → to compare categories and installs

***Scatter plots*** → to study relationships between variables

***Count plots*** → for sentiment analysis

***Key insights:***

* Most apps have ratings between 3.5 and 4.5

* Free apps have higher installs than paid apps

* Categories like Games and Communication dominate

* Most user reviews are positive

***Visualization 1: Distribution of App Ratings***

```python
#Histogram plot is used to show the distribution of numerical data
#To understand how Play Store apps are rated overall and identify common rating ranges.

plt.figure(figsize=(8,5))
sns.histplot(apps["Rating"], bins=20, kde=True)
plt.title("Distribution of App Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Apps")
plt.show()
```

***Visualization 2: Bar Plot: Top 10 Categories by Total Installs***

```python

#To identify the most popular and high-demand app categories.
top_categories = apps.groupby("Category")["Installs"].sum().sort_values(ascending=False).head(10)
#A barplot is used to compare values across different categories
plt.figure(figsize=(10,5))
sns.barplot(x=top_categories.index, y=top_categories.values)
plt.xticks(rotation=75)
plt.title("Top 10 App Categories by Installs")
plt.xlabel("Category")
plt.ylabel("Total Installs")
plt.show()
```

***Visualization 3: Free vs Paid Apps – Average Rating***

```python
# A barplot is used to compare values across different categories
#To compare user satisfaction between free and paid apps.
plt.figure(figsize=(6,4))
sns.barplot(x="Type", y="Rating", data=apps)
plt.title("Average Rating: Free vs Paid Apps")
plt.xlabel("App Type")
plt.ylabel("Average Rating")
plt.show()
```

***Visualization 4: App Size vs Rating***

```python
#A scatterplot is used to show the relationship between two numerical variables
  #1)Each point represents one data record
  #2)Helps identify patterns, trends, and correlations

#To check whether app size affects user ratings.
plt.figure(figsize=(7,5))
sns.scatterplot(x="Size", y="Rating", data=apps)
plt.title("App Size vs Rating")
plt.xlabel("App Size (MB)")
plt.ylabel("Rating")
plt.show()
```

***Visualization 5: Most Reviewed Apps***

```python
# To identify apps with the highest user engagement.
top_reviews = apps.sort_values(by="Reviews", ascending=False).head(10)

plt.figure(figsize=(10,5))
sns.barplot(x=top_reviews["App"], y=top_reviews["Reviews"])
plt.xticks(rotation=75)
plt.title("Top 10 Most Reviewed Apps")
plt.xlabel("App")
plt.ylabel("Number of Reviews")
plt.show()
```

***Visualization 6: User Sentiment Distribution (Reviews Dataset)***

```python
#A countplot is used to count the number of occurrences in each category
#To understand overall customer sentiment towards apps.
plt.figure(figsize=(6,4))
sns.countplot(x="Sentiment", data=reviews)
plt.title("User Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.show()
```
